```python
import zipfile
import os

zip_path = "/content/Elevvo Internship Tasks 10.zip"
extract_path = "/content/task10"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("Project extracted successfully!")
```

```python
import os

for root, dirs, files_list in os.walk(extract_path):
    level = root.replace(extract_path, "").count(os.sep)
    if level > 2:
        dirs[:] = []
        continue

    print("  " * level + os.path.basename(root) + "/")

    for file in files_list:
        print("  " * (level + 1) + file)
```

```python
import os

for root, dirs, files_list in os.walk("/content/task10"):
    if "app.py" in files_list and "requirements.txt" in files_list:
        PROJECT_DIR = root
        break
else:
    raise FileNotFoundError(
        "Could not find app.py and requirements.txt together."
    )

os.chdir(PROJECT_DIR)

print("Project directory:", PROJECT_DIR)
print("Current directory:", os.getcwd())
```

```python
from google.colab import userdata
import os

try:
    key = userdata.get("OPENAI_API_KEY")
    if key:
        os.environ["OPENAI_API_KEY"] = key
        print("API key found and loaded.")
    else:
        print("API key is empty.")
except Exception as e:
    print("Could not read Colab Secret:", type(e).__name__)
```

```python
from openai import OpenAI

client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])

try:
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[{"role": "user", "content": "Reply with OK."}],
        max_tokens=5
    )
    print("OpenAI connection successful!")
    print("Response:", response.choices[0].message.content)
except Exception as e:
    print("API error:", type(e).__name__)
    print(str(e)[:1000])
```
